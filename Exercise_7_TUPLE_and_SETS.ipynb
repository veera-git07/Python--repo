```python
#set operation in python
#using operators like,|,^....etc
mySet={1,2,3,4,5,6,7}
yourSet={3,4,5,6,2,1,21,12,1}
print(mySet | yourSet) # Union operation using "|"
print(mySet & yourSet)# Intersection operation using "&"
print(mySet - yourSet)# Difference operation using "-"
print(mySet ^ yourSet) # Symmetric Difference operation using "^"
print(mySet)
```

```python
# Python Program to demonstrate Set Methods

A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

print("Set A:", A)
print("Set B:", B)
A.add(10)
print("After add(10):", A)
A.remove(10)
print("After remove(10):", A)
A.discard(5)
print("After discard(5):", A)
print("Union:", A.union(B))
print("Intersection:", A.intersection(B))
print("Difference (A - B):", A.difference(B))
print("Symmetric Difference:", A.symmetric_difference(B))
C = {1, 2}
print("Is C a subset of A", C.issubset(A))
print("Is A a superset of C", A.issuperset(C))
D = {20, 30}
print("Are A and D disjoint", A.isdisjoint(D))
E = A.copy()
print("Copy of A:", E)
E.clear()
print("After clear():", E)
```
