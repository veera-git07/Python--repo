```python
#reverse of a string:
def reverse_str(s):
  rev=""
  for i in range(len(s)-1,-1,-1):
    rev=rev+s[i]
  return rev
print(reverse_str("Veera"))
```

```python
#default arguments :
def print_msg(name,branch="AI&DS"):
  print("The name is",name," and the branch is",branch)
print_msg("sai")
print_msg("Veera","CSE")
```

```python

#keyword arguments:
def print_msg(name,branch):
  print("The name is",name,"and the branch is",branch)
print_msg(name="Praveen",branch="CSE")
print_msg(branch="AI&DS",name="Veera")
```

```python
#positional arguments:
def print_msg(name,branch):
   print("The name is",name,"and the branch is",branch)
print_msg("CSE","Yuva")
print_msg("Veera","AI&DS")
```

```python
#check whether the given number is palindrome or not using python functions:
def palindrom_e(num):
  real=num
  rev=0
  while(num!=0):
    rev=rev*10+num%10
    num= num//10
  if(rev==real):
      print("PALINDROME")
  else:
      print("NOTE A PALINDROME")
palindrom_e(123321)
palindrom_e(143)
```
