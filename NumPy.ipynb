```python
#to add two matrices
r = int(input("Enter number of rows: "))
c = int(input("Enter number of columns: "))
print("Enter first matrix:")
A = []
for i in range(r):
    row = []
    for j in range(c):
        row.append(int(input()))
    A.append(row)
print("Enter second matrix:")
B = []
for i in range(r):
    row = []
    for j in range(c):
        row.append(int(input()))
    B.append(row)
C = []
for i in range(r):
    row = []
    for j in range(c):
        row.append(A[i][j] + B[i][j])
    C.append(row)
print("Sum of two matrices:")
for row in C:
    print(row)
```

```python
# matrix multiplication
r1 = int(input("Enter rows of first matrix: "))
c1 = int(input("Enter columns of first matrix: "))
r2 = int(input("Enter rows of second matrix: "))
c2 = int(input("Enter columns of second matrix: "))
if c1 != r2:
    print("Matrix multiplication is not possible")
else:
    print("Enter first matrix:")
    A = []
    for i in range(r1):
        row = []
        for j in range(c1):
            row.append(int(input()))
        A.append(row)
    print("Enter second matrix:")
    B = []
    for i in range(r2):
        row = []
        for j in range(c2):
            row.append(int(input()))
        B.append(row)
    C = []
    for i in range(r1):
        row = []
        for j in range(c2):
            sum = 0
            for k in range(c1):
                sum = sum + A[i][k] * B[k][j]
            row.append(sum)
        C.append(row)
    print("Product of two matrices:")
    for row in C:
        print(row)
```

```python
# to find transpose of a matrix
r = int(input("Enter number of rows: "))
c = int(input("Enter number of columns: "))
print("Enter matrix elements:")
A = []

for i in range(r):
    row = []
    for j in range(c):
        row.append(int(input()))
    A.append(row)
T = []

for i in range(c):
    row = []
    for j in range(r):
        row.append(A[j][i])
    T.append(row)
print("Original Matrix:")
for row in A:
    print(row)
print("Transpose Matrix:")
for row in T:
    print(row)
```
