```python

# Opening source file in read mode
f1 = open("source.txt", "r")

# Reading the content
data = f1.read()

# Converting all words into lowercase
data = data.lower()

# Splitting the content into words
words = data.split()

# Sorting the words
words.sort()

# Opening destination file in write mode
f2 = open("destination.txt", "w")

# Writing sorted words into destination file
for word in words:
    f2.write(word + "\n")

# Closing both files
f1.close()
f2.close()

print("Words sorted and copied successfully")
```

```python

# Reading the order of matrices
r = int(input("Enter number of rows: "))
c = int(input("Enter number of columns: "))

# Reading first matrix
print("Enter first matrix:")
A = []
for i in range(r):
    row = []
    for j in range(c):
        row.append(int(input()))
    A.append(row)

# Reading second matrix
print("Enter second matrix:")
B = []
for i in range(r):
    row = []
    for j in range(c):
        row.append(int(input()))
    B.append(row)

# Adding two matrices
C = []
for i in range(r):
    row = []
    for j in range(c):
        row.append(A[i][j] + B[i][j])
    C.append(row)

# Displaying result
print("Sum of two matrices:")
for row in C:
    print(row)
```

```python

import pandas as pd

# Creating a dictionary with five keys
data = {
    'Name': ['Ram', 'Ravi', 'Krishna', 'John', 'David',
             'Sam', 'Raj', 'Kiran', 'Hari', 'Arun'],

    'Age': [20, 21, 19, 22, 20, 21, 23, 19, 22, 20],

    'Marks': [85, 90, 78, 88, 92, 76, 89, 95, 80, 87],

    'City': ['Hyderabad', 'Vijayawada', 'Chennai', 'Delhi', 'Mumbai',
             'Vizag', 'Bangalore', 'Pune', 'Kolkata', 'Guntur'],

    'Grade': ['A', 'A', 'B', 'A', 'A', 'B', 'A', 'A', 'B', 'A']
}

# Converting dictionary into DataFrame
df = pd.DataFrame(data)

# Displaying complete DataFrame
print("Complete DataFrame:")
print(df)

# a) Applying head() function
print("\nFirst five rows:")
print(df.head())

# b) Data selection operations

# Selecting a single column
print("\nSelecting Name column:")
print(df['Name'])

# Selecting multiple columns
print("\nSelecting Name and Marks columns:")
print(df[['Name', 'Marks']])

# Selecting rows using iloc
print("\nSelecting first three rows:")
print(df.iloc[0:3])

# Selecting rows using loc
print("\nSelecting rows from index 0 to 2:")
print(df.loc[0:2])

# Selecting data based on condition
print("\nStudents having marks greater than 85:")
print(df[df['Marks'] > 85])
```
